#####PROBLEM 1

```python
# car['year'].str.isnumeric() #check ki year pr ek string operation karo jo chk kare ki data numeric hai? har row pr
# car[car['year'].str.isnumeric()] #boolean toh it can go inside car toh vahi rows filter hongi jo true hai
car=car[car['year'].str.isnumeric()] #store it in car
```

#####PROBLEM 2

```python
car['year']=car['year'].astype(int) #datatype changed to int and storedin the old column itself
```

short check --> done correctly

```python
car.info()
```

#####PROBLEM 3

```python
car=car[car['Price']!="Ask For Price"] #take whose value is not ask for price and store it back in car
```

#####PROBLEM 4

```python
#replace commas by empty string
#then convert whole string into int
car['Price']=car['Price'].str.replace(',','').astype(int)
```

```python
car.info()
```

#####PROBLEM 5 & 6

```python
#split the string by space
#and uska pehla element ham rakh lenge--> eg 45000 km --> [45000,kms]--> mei se 45000 rakh liya
#replace commas with empty string
car['kms_driven']=car['kms_driven'].str.split(' ').str.get(0).str.replace(',','')
```

```python
car=car[car['kms_driven'].str.isnumeric()] #removing petrol and nan rows as not numeric
```

```python
#converting them into int datatype
car['kms_driven']=car['kms_driven'].astype(int)
```

```python
car.info()
```

#####PROBLEM 7

```python
#Nahi, .isna() ke saath .str nahi lagana chahiye
#.str ek accessor hai jo sirf tab use hota hai jab tu string-specific operations karni ho jaise text ke andar kuch check karna (isnumeric(), contains(), upper(), lower(), etc.)
#.isna()-->.isna() check karta hai ki value missing/null (NaN) hai ya nahi yeh koi text operation nahi hai, balki ek general check hai jo kisi bhi data type pe kaam karta hai (numbers ho, text ho, dates ho, kuch bhi ho). Isiliye ismein .str ki zaroorat hi nahi pandas seedha values check kar leta hai ki woh NaN/None/missing hai ya nahi, without needing to treat them as strings.
#~ this means excluding all the rows jisne voh lag raha
car=car[~car['fuel_type'].isna()];
```

```python
car.info()
```

#####PROBLEM 8

```python
#split using space
#then slice it--> 0 se lekr 3--> (0,3)--> 0 included 3 excluded
#take first 3
#the join these words using space
car['name']=car['name'].str.split(' ').str.slice(0,3).str.join(' ')
```

```python
#now indexes have been unordered--> so reset them
#Bina drop=True ke, reset_index() purane (gaps wale) index ko ek naya column bana ke dataframe mein daal deta (jispe naam hota index) taki original numbering safe rahe
#drop=True likhne se pandas ko bol rahe hain: "purana index bilkul phenk do (drop), koi naya column mat banao"
car=car.reset_index(drop=True)
```
