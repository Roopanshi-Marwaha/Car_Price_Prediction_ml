#EXTRACT FEATURES AND LABELS

````python
#everything is our feature input except price which is output
X=car.drop(columns='Price') #all inputs
Y=car['Price'] #output
````

````python
# from sklearn.model_selection import train_test_split
# sklearn (scikit-learn) — Python ki sabse popular ML library, jisme ready-made models aur tools hote hain
# model_selection — sklearn ke andar ek module hai jo data split karne, model evaluate karne jaise cheezon ke liye hai
# train_test_split — ek function hai jo data ko do parts mein baant deta hai: training ke liye aur testing ke liye

# X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=0.2)
# Right side pehle samajh: train_test_split(X, Y, test_size=0.2)
# X — tere features/independent variables (jaise company, year, kms_driven, fuel_type — jo cheezein price predict karne ke liye use hongi)
# Y — tera target/dependent variable (jo predict karna hai — yaha Price)
# test_size=0.2 — 20% data testing ke liye rakho, baaki 80% training ke liye
# Left side — function 4 cheezein return karta hai (isliye 4 variables mein store ho raha hai):
# X_train — 80% features, model train karne ke liye
# X_test — 20% features, model test karne ke liye
# Y_train — 80% ke corresponding actual prices, training ke liye
# Y_test — 20% ke corresponding actual prices, testing ke liye

from sklearn.model_selection import train_test_split
X_train,X_test,Y_train,Y_test=train_test_split(X,Y,test_size=0.2)
````

````python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score
#r2_score yeh function R² score (R-squared) calculate karta hai, jo batata hai model ne data ko kitna acha fit kiya. Value 0 se 1 ke beech hoti hai (1 ke jitna paas, utna acha model) regression models ke liye standard evaluation metric hai. Yeh training ke baad Y_test (actual) vs prediction compare karne ke liye use hoga.
from sklearn.preprocessing import OneHotEncoder
#yeh categorical (text) columns ko numbers mein convert karta hai, kyunki ML models sirf numbers samajhte hain, text nahi. Tere dataset mein company, fuel_type jaise columns text hain model inhe directly use nahi kar sakta.
from sklearn.compose import make_column_transformer
from sklearn.pipeline import make_pipeline
````

````python
ohe=OneHotEncoder() #making an obj of ohe
ohe.fit(X[['name','company','fuel_type']]) #putting the categorical values of X in []
````

````python
# ohe.categories_ #pass this to ohe
````

````python
column_trans = make_column_transformer(
    (OneHotEncoder(categories=ohe.categories_), ['name', 'company', 'fuel_type']),
    remainder='passthrough'
)
#hamare data mein kuch columns text wale hain--> Linear Regression directly text nahi samajh sakta--> so this column_trans pehle in columns ko numbers mein convert karega using One Hot Encoding.
#Conversion ke baad jo numerical data mila, woh LinearRegression ko jayega.
#tranform columns using this and applying ohe on the given columns only baki passtrough
#ohe mei categories bhejni hoti hai
````

````python
lr = LinearRegression()
````

````python
pipe = make_pipeline(column_trans, lr)
#jo ham input data data hai pipeline ko ohe lagega voh sabh columns transform hongi
#abh yeh tranformd + previous columns lrm ko diye jayenge
#raw data apne aap tranform ho gaya
#make_pipeline --> Aur iska kaam hai multiple steps ko ek pipeline mein jod dena.
````

````python
pipe.fit(X_train, Y_train)
#we don't need to manually do it-->Pipeline ye steps automatically order mein karwa degi.
#"Pehle data ko transform karega phir Linear Regression lagayega
#.fit() → un steps ko training data par chala kar model ko train karta hai
````

````python
y_pred=pipe.predict(X_test)
#Jo model humne X_train, Y_train se train kiya tha,
#ab usko X_test do aur bolo: in cars ki Price predict karo.
````

````python
y_pred
````

````python
r2_score(Y_test, y_pred)
````

````python
scores = []

for i in range(1000):

    X_train, X_test, Y_train, Y_test = train_test_split(
        X, Y, test_size=0.2, random_state=i
    )

    lr = LinearRegression()

    pipe = make_pipeline(column_trans, lr)

    pipe.fit(X_train, Y_train)

    Y_pred = pipe.predict(X_test)

    print(r2_score(Y_test, Y_pred), i)

    scores.append(r2_score(Y_test, Y_pred))

#Jo R² score abhi mila hai, usko scores wali list mein save kar do.
#Har different train-test split ka R² score hum save kar rahe hain.
#Aur ye save karna zaroori hai kyunki last mein hum in 1000 scores mein se best score dhoondhne wale hain.
````

````python
np.argmax(scores)
#calculating max R2 index-->np.argmax() maximum value ka index return karta hai.
````

# Train-Test Split — Simple Car Example

Let’s understand `X_train`, `X_test`, `Y_train`, and `Y_test` using an actual car example.

Suppose you have **5 cars in your original dataset**:

| Car   | Year | Kms | Fuel   | Price |
| ----- | ---: | --: | ------ | ----: |
| Swift | 2015 | 50k | Petrol |    4L |
| i20   | 2018 | 30k | Petrol |    6L |
| Alto  | 2014 | 70k | Petrol |  2.5L |
| Creta | 2019 | 20k | Diesel |   12L |
| City  | 2017 | 40k | Petrol |    8L |

---

## 1. First, we create X and Y

We write:

```python
X = car.drop(columns='Price')
Y = car['Price']
```

### X

```text
Swift   2015   50k   Petrol
i20     2018   30k   Petrol
Alto    2014   70k   Petrol
Creta   2019   20k   Diesel
City    2017   40k   Petrol
```

These are the **inputs/features**.

### Y

```text
4L
6L
2.5L
12L
8L
```

These are the **answers/target values (Price)**.

---

# 2. Now we perform train-test split

Suppose we write:

```python
test_size=0.2
```

This means approximately **20% of the data goes to testing** and the remaining **80% goes to training**.

Suppose randomly **City** goes into the testing data.

Then:

### X_train

```text
Swift   2015   50k   Petrol
i20     2018   30k   Petrol
Alto    2014   70k   Petrol
Creta   2019   20k   Diesel
```

### Y_train

```text
4L
6L
2.5L
12L
```

Notice:

**X_train and Y_train contain corresponding information for the same cars.**

```text
X_train                         Y_train

Swift, 2015, 50k, Petrol   →      4L
i20, 2018, 30k, Petrol     →      6L
Alto, 2014, 70k, Petrol    →      2.5L
Creta, 2019, 20k, Diesel   →      12L
```

We use these **4 cars to train the model**.

---

# 3. What is X_test?

The car that did not go into the training data goes into the testing data.

```text
X_test

City   2017   40k   Petrol
```

And its actual price is:

```text
Y_test

8L
```

**Important:**

During training, the model does **not see City or its actual price of 8L**.

---

# 4. Now `.fit()` happens

```python
pipe.fit(X_train, Y_train)
```

This means:

> "Model, look at these 4 cars and their prices and learn the relationship between the inputs and the price."

The model sees:

```text
Swift → 4L
i20   → 6L
Alto  → 2.5L
Creta → 12L
```

From these examples, the model learns the relationship between the features and the price.

---

# 5. Now comes the model's exam

We write:

```python
Y_pred = pipe.predict(X_test)
```

We give the model **only `X_test`**:

```text
City   2017   40k   Petrol
```

And ask:

> "What do you think this car's price is?"

The model might predict:

```text
Y_pred = 7.5L
```

But we already know the actual price:

```text
Y_test = 8L
```

---

# 6. Now compare actual vs predicted

```text
Actual price:       8L
Model prediction:   7.5L
```

Then:

```python
r2_score(Y_test, Y_pred)
```

checks how well the model performed on the testing data.

---

# The simplest explanation

### `X_train`

**Information about the cars used for training**

```text
Swift, i20, Alto, Creta...
```

### `Y_train`

**Actual prices of those training cars**

```text
4L, 6L, 2.5L, 12L...
```

### `X_test`

**Information about the cars used for testing**

```text
City, 2017, 40k, Petrol
```

### `Y_test`

**Actual price of the testing car**

```text
8L
```

---

# Complete Flow

```text
                 ORIGINAL DATA
                       ↓
                train_test_split
                       ↓
             ┌─────────┴─────────┐
             ↓                   ↓
         TRAINING             TESTING
             ↓                   ↓
     X_train + Y_train         X_test
             ↓                   ↓
           .fit()            .predict()
             ↓                   ↓
        MODEL LEARNS           Y_pred
                                 ↓
                          Y_pred vs Y_test
                                 ↓
                              R² Score
```

### The most important thing to remember:

> **X = Questions**

> **Y = Answers**

> **Train = Give the model questions + answers so it can learn.**

> **Test = Give the model only the questions, get its answers, and compare them with the actual answers.**

That is the whole idea behind `X_train`, `Y_train`, `X_test`, and `Y_test`.
